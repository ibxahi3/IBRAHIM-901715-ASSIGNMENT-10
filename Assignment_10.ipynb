## Question 12 — Electricity Bill Function

**Problem:**  
An electricity company charges different rates based on units consumed. Create a function that calculates the bill.

**Suggested rates (you may adjust slightly):**
- 0 – 100 units → Rs. 5 per unit
- 101 – 200 units → Rs. 10 per unit
- Above 200 units → Rs. 15 per unit

**Requirements:**
- Create a function that accepts the number of **units consumed**.
- Calculate and **return** the final electricity bill using conditions.
- Display the returned bill outside the function.

**Focus:** Function + arguments + conditions + `return`

````python
# Write your code below

def electricity_bill(units):

    if units <= 100:
        bill = units * 5

    elif units <= 200:
        bill = (100 * 5) + ((units - 100) * 10)

    else:
        bill = (100 * 5) + (100 * 10) + ((units - 200) * 15)

    return bill


total_bill = electricity_bill(250)

print("Electricity Bill: Rs.", total_bill)
````

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

````python
# Write your code below

def marks_processor(marks):
    highest = marks[0]
    lowest = marks[0]
    total = 0

    for mark in marks:
        total += mark

        if mark > highest:
            highest = mark

        if mark < lowest:
            lowest = mark

    average = total / len(marks)

    return highest, lowest, total, average


student_marks = [85, 90, 75, 60, 95]

highest, lowest, total, average = marks_processor(student_marks)

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Total Marks:", total)
print("Average Marks:", average)
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below

def process_order(price, quantity):

    def calculate_subtotal():
        return price * quantity

    def calculate_discount():
        return calculate_subtotal() * 0.10

    def calculate_tax():
        return (calculate_subtotal() - calculate_discount()) * 0.05

    def calculate_final_amount():
        subtotal = calculate_subtotal()
        discount = calculate_discount()
        tax = calculate_tax()

        return subtotal - discount + tax

    return calculate_final_amount()


result = process_order(1000, 5)

print("Final Order Amount:", result)
````

## Question 28 — Smart Shopping Billing System

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**
- Allow the user (or hard-code) multiple products.
- Calculate:
  - Individual product totals
  - Overall subtotal
  - Discount
  - Tax
  - Final bill
- Use **multiple functions** to separate responsibilities.
- At least one function must use `*args` or a flexible-argument design.
- Display a professional-looking bill.

**Focus:** Real-world application + multiple functions + logic building

````python
# Write your code below

def product_total(price, quantity):
    return price * quantity


def calculate_subtotal(*totals):
    return sum(totals)


def calculate_discount(subtotal):
    if subtotal >= 5000:
        return subtotal * 0.10
    else:
        return 0


def calculate_tax(amount):
    return amount * 0.05


def final_bill(subtotal, discount, tax):
    return subtotal - discount + tax


# Products
products = [
    {"name": "Laptop", "price": 50000, "quantity": 1},
    {"name": "Mouse", "price": 1000, "quantity": 2},
    {"name": "Keyboard", "price": 2000, "quantity": 1}
]

totals = []

print("========== SHOPPING BILL ==========")

for product in products:
    total = product_total(
        product["price"],
        product["quantity"]
    )

    totals.append(total)

    print(product["name"], ":", total)

subtotal = calculate_subtotal(*totals)
discount = calculate_discount(subtotal)
tax = calculate_tax(subtotal - discount)
final = final_bill(subtotal, discount, tax)

print("-----------------------------------")
print("Subtotal:", subtotal)
print("Discount:", discount)
print("Tax:", tax)
print("Final Bill:", final)
print("===================================")
````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
# Write your code below

def student_info(name, roll_no):
    return name, roll_no


def calculate_total(marks):
    total = 0

    for mark in marks:
        total += mark

    return total


def calculate_percentage(total, subjects):
    return total / subjects


def calculate_grade(percentage):
    if percentage >= 80:
        return "A+"
    elif percentage >= 70:
        return "A"
    elif percentage >= 60:
        return "B"
    elif percentage >= 50:
        return "C"
    else:
        return "F"


def check_result(marks):
    for mark in marks:
        if mark < 40:
            return "Fail"

    return "Pass"


def display_result(name, roll_no, marks):

    total = calculate_total(marks)
    percentage = calculate_percentage(total, len(marks))
    grade = calculate_grade(percentage)
    result = check_result(marks)

    print("\n====== STUDENT RESULT CARD ======")
    print("Name:", name)
    print("Roll Number:", roll_no)
    print("Subject Marks:", marks)
    print("Total Marks:", total)
    print("Percentage:", percentage, "%")
    print("Grade:", grade)
    print("Result:", result)
    print("=================================")


# Student information
name, roll_no = student_info("Ibrahim", 101)

# Subject marks
marks = [85, 75, 90, 80, 70]

display_result(name, roll_no, marks)
````

````python
# Write your code below

# Customer information
def customer_info(**kwargs):
    return kwargs


# Product selection
def select_products():
    products = [
        {"name": "Laptop", "price": 50000, "quantity": 1},
        {"name": "Mouse", "price": 1000, "quantity": 2},
        {"name": "Keyboard", "price": 2000, "quantity": 1}
    ]

    return products


# Calculate individual product totals
def calculate_products(products):

    totals = []

    for product in products:
        total = product["price"] * product["quantity"]

        totals.append(total)

    return totals


# Calculate subtotal using *args
def calculate_subtotal(*totals):
    return sum(totals)


# Calculate discount using nested function
def calculate_discount(subtotal):

    def discount_rate():
        if subtotal >= 50000:
            return 0.10
        elif subtotal >= 20000:
            return 0.05
        else:
            return 0

    return subtotal * discount_rate()


# Calculate tax
def calculate_tax(amount, tax_rate=0.05):
    return amount * tax_rate


# Delivery charges
def delivery_charges(subtotal):
    if subtotal >= 50000:
        return 0
    else:
        return 300


# Final bill calculation
def calculate_final_bill(subtotal, discount, tax, delivery):
    return subtotal - discount + tax + delivery


# Display order summary
def display_order(customer, products, subtotal, discount, tax, delivery, final):

    print("\n========== ORDER SUMMARY ==========")

    print("Customer:", customer["name"])
    print("City:", customer["city"])

    print("\n---------- PRODUCTS ----------")

    for product in products:
        total = product["price"] * product["quantity"]

        print(
            product["name"],
            "| Quantity:", product["quantity"],
            "| Total:", total
        )

    print("-----------------------------------")
    print("Subtotal:", subtotal)
    print("Discount:", discount)
    print("Tax:", tax)
    print("Delivery:", delivery)
    print("-----------------------------------")
    print("FINAL AMOUNT:", final)
    print("===================================")


# Main program

# Customer details
customer = customer_info(
    name="Ibrahim",
    city="Karachi"
)

# Get products
products = select_products()

# Calculate product totals
totals = calculate_products(products)

# Lambda to check whether order qualifies for free delivery
free_delivery = lambda amount: amount >= 50000

# Calculate bill
subtotal = calculate_subtotal(*totals)

discount = calculate_discount(subtotal)

tax = calculate_tax(subtotal - discount)

delivery = delivery_charges(subtotal)

final = calculate_final_bill(
    subtotal,
    discount,
    tax,
    delivery
)

# Display final order
display_order(
    customer,
    products,
    subtotal,
    discount,
    tax,
    delivery,
    final
)

print("Free Delivery Eligible:", free_delivery(subtotal))
````
